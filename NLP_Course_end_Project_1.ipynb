## Task 2
Normalize casings for the review text and extract the text into a list for easier manipulation.

```python
# Normalize casings
review_df['review'] = review_df['review'].str.lower()

# Extract reviews into a list
reviews_list = review_df['review'].tolist()

# Display the first few normalized reviews to verify
print("First 5 normalized reviews:")
for i in range(min(5, len(reviews_list))):
    print(reviews_list[i])
```

## Task 3
Tokenize the reviews using NLTKs `word_tokenize` function.

```python
import nltk
import re
from nltk.tokenize import word_tokenize

# Download the 'punkt' tokenizer if not already downloaded
nltk.download('punkt')

# Clean the reviews
cleaned_reviews = [re.sub(r'[^a-zA-Z0-9\s]', '', review) for review in reviews_list]

# tokenize the cleaned reviews
tokenized_reviews = [word_tokenize(review) for review in cleaned_reviews]

print("First 5 tokenized reviews:")
for i in range(min(5, len(tokenized_reviews))):
    print(tokenized_reviews[i])
```

## Task 4
Perform parts-of-speech tagging on each sentence using the NLTK POS tagger.

```python
from nltk import pos_tag
import nltk

nltk.download('averaged_perceptron_tagger_eng')


# Perform POS tagging on each tokenized review
pos_tagged_reviews = [pos_tag(review) for review in tokenized_reviews]

# Display the first few POS-tagged reviews to verify
print("First 5 POS-tagged reviews:")
for i in range(min(5, len(pos_tagged_reviews))):
    print(pos_tagged_reviews[i])
```

## Task 5

For the topic model, we should  want to include only nouns.

1.   Find out all the POS tags that correspond to nouns.
2.   Limit the data to only terms with these tags.







```python
# Define the noun tags
noun_tags = ['NN', 'NNS', 'NNP', 'NNPS']

# Filter the POS tagged reviews to keep only nouns
noun_reviews = []

for review in pos_tagged_reviews:
    nouns = [word for (word, tag) in review if tag in noun_tags]
    noun_reviews.append(nouns)

for i in range(min(5, len(noun_reviews))):
    print(noun_reviews[i])
```

## Task 6

1.   Different forms of the terms need to be treated as one.
2.   No need to provide POS tag to lemmatizer for now.

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Download the 'wordnet' and 'omw-1.4' resources if not already downloaded
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize wordnet lemmatizer
wnl = WordNetLemmatizer()

for i in range(len(noun_reviews)):
    for j in range(len(noun_reviews[i])):
        noun_reviews[i][j] = wnl.lemmatize(noun_reviews[i][j])

# Display the first few lemmatized reviews to verify
print("\nFirst 5 lemmatized reviews with only nouns:")
for i in range(min(5, len(noun_reviews))):
    print(noun_reviews[i])
```

## Task 7
Remove stopwords and punctuation (if there are any).

```python
# Remove stopwords and punctuation (if there are any).
import string
from nltk.corpus import stopwords

# Download the 'stopwords' resource if not already downloaded
try:
    nltk.data.find('corpora/stopwords')
except nltk.downloader.DownloadError:
    nltk.download('stopwords')


stop_words = set(stopwords.words('english'))

stoped_reviews = []

for review in noun_reviews:
    filtered_review = [word for word in review if word not in stop_words and word not in string.punctuation]
    stoped_reviews.append(filtered_review)

# Print
for i in range(min(5, len(stoped_reviews))):
    print(stoped_reviews[i])
```

## Task 8
Create a topic model using LDA on the cleaned-up data with 12 topics.

1.   Print out the top terms for each topic.
2.   What is the coherence of the model with the c_v metric?

```python
# Install gensim
try :
   import gensim
   print("Gensim imported successfully!")
except ModuleNotFoundError:
   print("Gensim not found.")
   %pip install gensim
   import gensim
   print("Gensim installed successfully!")
```

```python
from gensim import corpora

# Create a dictionary from the processed reviews
dictionary = corpora.Dictionary(stoped_reviews)

# Create a corpus from the dictionary
corpus = [dictionary.doc2bow(review) for review in stoped_reviews]

print("First 5 elements of the corpus:")
for i in range(min(5, len(corpus))):
    print(corpus[i])
```

Next, will build the LDA topic model with 12 topics.

```python
from gensim.models import LdaModel

# Build the LDA model
num_topics = 12
lda_model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, random_state=42, passes=10)

# Print the top terms for each topic
print(f"\nTop terms for each of the {num_topics} topics:")
for topic in lda_model.print_topics(num_words=10):
    print(topic)
```

Finally, will calculate the coherence of the model using the c_v metric.

```python
from gensim.models.coherencemodel import CoherenceModel

# Calculate Coherence Score
coherence_model_lda = CoherenceModel(model=lda_model, texts=stoped_reviews, dictionary=dictionary, coherence='c_v')
coherence_lda = coherence_model_lda.get_coherence()

print(f'\nCoherence Score (c_v): {coherence_lda}')
```
